## 1. 📦 Data Warehouse vs Data Lake vs Data Lakehouse

| | Data Warehouse | Data Lake | Data Lakehouse |
|---|---|---|---|
| **Data Type** | Structured only | All types (raw, unstructured) | All types |
| **Schema** | Schema-on-write | Schema-on-read | Both supported |
| **Cost** | Expensive | Cheap storage | Cost-effective |
| **Performance** | Fast queries | Slow without tuning | Fast (Delta Lake) |
| **Use Case** | BI & Reporting | Data science, ML | BI + ML + Streaming |

### 🔑 Key Idea
> **Data Lakehouse = Data Lake + Data Warehouse**  
> Best of both worlds — cheap storage of a lake + the reliability & performance of a warehouse.

**Production Tip:** In most modern pipelines, you'll use a Lakehouse. Choose it when your team needs both analysts (SQL) and data scientists (ML) working on the same data — avoids data duplication between systems.

## 2. 🏅 Medallion Architecture

A layered data organization pattern used inside a Lakehouse.

```
Raw Data → [Bronze] → [Silver] → [Gold] → BI / ML
```

| Layer | What it holds | Think of it as |
|---|---|---|
| **Bronze** | Raw, unprocessed data as-is | Landing zone |
| **Silver** | Cleaned, filtered, joined data | Trusted data |
| **Gold** | Aggregated, business-ready data | Serving layer |

### Why Medallion? Key Benefits:

- **Data Lineage & Traceability** — You always know where data came from. If Gold breaks, trace back to Silver → Bronze. Critical for debugging in production.
- **Data Governance & Compliance** — Control who sees what at each layer (e.g., PII masked in Silver, removed in Gold). Helps with GDPR, HIPAA.
- **Incremental Processing** — Only process new/changed data, not full reloads. Saves cost and time at scale.
- **Performance & Scalability** — Each layer is optimized. Gold tables are small, indexed, and fast for dashboards.
- **Role-Based Access (Security)** — Data engineers own Bronze/Silver. Analysts only access Gold. Enforced via Unity Catalog.

**Production Tip:** Never let analysts query Bronze directly. Bronze is messy, incomplete, and schema can change anytime. Always expose Gold or well-curated Silver.

## 3. ⚡ Apache Spark — The Engine Under the Hood

> **"Lightning-fast unified analytics engine for big data processing and machine learning"**

### What makes Spark special:

- **In-memory processing** — Data stays in RAM across operations, not written to disk between steps → massive speed gains vs Hadoop MapReduce
- **Distributed computing** — Work is split across many machines (nodes) automatically
- **Unified engine** — One tool handles: SQL queries, Streaming, ML, Graph processing
- **100% Open Source** — Apache License, huge community
- **Simple APIs** — Python (PySpark), Scala, SQL, R — you pick your language

**Production Tip:** Spark shines when data is too big for a single machine (GBs to TBs+). For small data (<1GB), Spark is overkill — use Pandas or plain SQL.

## 4. 🔷 Databricks — Spark + Everything Else

> **Databricks = Apache Spark + managed platform built by Spark's creators**

The Spark creators founded Databricks to solve the pain of managing Spark yourself.

### What Databricks adds on top of Spark:

| Component | What it does | Why it matters in Production |
|---|---|---|
| **Clusters** | Managed Spark compute | Auto-scales up/down, saves cost |
| **Workspace / Notebooks** | Collaborative coding environment | Teams work together in one place |
| **Optimized Spark Engine** | Faster than open-source Spark | Better query plans out of the box |
| **Photon** | C++ vectorized query engine | 2-5x faster SQL queries, especially for heavy analytics |
| **Delta Lake** | Storage layer with ACID transactions | Reliable, no corrupted data, supports updates/deletes |
| **Unity Catalog** | Centralized governance & access control | One place to manage permissions across all data assets |
| **Delta Live Tables (DLT)** | Declarative pipeline framework | Build Bronze→Silver→Gold pipelines easily, auto data quality checks |
| **Workflows** | Job scheduler & orchestration | Schedule pipelines, handle dependencies, retry on failure |
| **SQL Warehouse** | Serverless SQL endpoint for analysts | Analysts query data without touching clusters |
| **MLflow** | ML experiment tracking & model registry | Track model versions, deploy models to production |
| **DatabricksIQ** | AI assistant inside Databricks | Helps write code, explain pipelines, debug |

### Cloud Storage Support:
- AWS S3, Azure ADLS (Data Lake Storage), Google Cloud Storage (GCS)

## 5. 🏗️ Databricks Architecture

```
┌─────────────────────────────────────────┐
│           CONTROL PLANE                 │
│  (Managed by Databricks)                │
│  - Web UI / Workspace                   │
│  - Job Scheduler                        │
│  - Cluster Manager                      │
│  - Unity Catalog                        │
└─────────────────────────────────────────┘
                    ↕
┌─────────────────────────────────────────┐
│           COMPUTE PLANE                 │
│  (Runs in YOUR cloud account)           │
│                                         │
│  ┌───────────────┐  ┌────────────────┐  │
│  │ Classic       │  │ Serverless     │  │
│  │ Compute Plane │  │ Compute Plane  │  │
│  │ (your VMs)    │  │ (instant start)│  │
│  └───────────────┘  └────────────────┘  │
└─────────────────────────────────────────┘
                    ↕
┌─────────────────────────────────────────┐
│        WORKSPACE CLOUD STORAGE          │
│  (S3 / ADLS / GCS — your cloud)         │
│  - Delta Tables                         │
│  - Notebooks, Libraries                 │
└─────────────────────────────────────────┘
```

### Control Plane vs Compute Plane — Key Distinction:

| | Control Plane | Compute Plane |
|---|---|---|
| **Who manages it** | Databricks Inc. | You (your cloud account) |
| **What's here** | UI, metadata, job configs | Actual Spark clusters, your data |
| **Data leaves?** | Only metadata/configs | Your data NEVER leaves your cloud |

**Production Tip:** This split is important for **security & compliance**. Your actual data stays in your cloud account (compute plane). Databricks only manages the orchestration layer (control plane). This is a common question in enterprise security reviews.

### Classic vs Serverless Compute:
- **Classic** — You provision clusters, you pay for idle time, more control
- **Serverless** — Instant start, no cluster management, pay per use. Best for SQL Warehouses and DLT pipelines
