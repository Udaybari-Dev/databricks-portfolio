```sql
%python
import zipfile
import os

# Your zip file path
zip_path = "/Volumes/circuitbox/landing/operational_data/circuitbox_data.zip"

# Extract to same volume
extract_path = "/Volumes/circuitbox/landing/operational_data/"

# Unzip
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_path)

print("✅ Unzipped successfully!")

# Verify - see all files
for root, dirs, files in os.walk(extract_path):
    for file in files:
        print(os.path.join(root, file))
```
