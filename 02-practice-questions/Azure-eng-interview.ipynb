

# Data Engineering Interview Preparation Questions

## 🔹 Basic / Intro Questions

1. Tell me about yourself.
2. Explain your current project in detail.
3. What architecture are you using in your project?
4. What are the different data sources in your pipeline?

---

## 🔹 Spark & Databricks Questions

5. Why are DataFrames immutable in Spark?
6. Can we write Java code in Databricks notebooks?
7. What are jobs in Spark (job1, job2, etc.)?
8. What is the relationship between jobs and stages?
9. How many jobs will be created for `select`, `filter`, and `show` operations?
10. What is the difference between transformations and actions?
11. How many transformations and actions are there in a given Spark code snippet?
12. Why does Spark separate transformations and actions?
13. What is DAG in Spark?
14. How does Spark execute a job internally?

---

## 🔹 Data Writing & Storage

15. How do you write a DataFrame to storage?
16. What is append mode vs overwrite mode?
17. In append mode, does Spark modify existing files or create new files?
18. Where do you store processed data in medallion architecture?

---

## 🔹 Delta Lake & Tables

19. What is a Delta table?
20. How is Delta table different from a normal Spark table?
21. What are the advantages of Delta Lake?
22. What features does Delta provide (like DML, ACID, etc.)?

## 🔹 CI/CD & Deployment
